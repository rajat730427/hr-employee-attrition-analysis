```python
sns.countplot(x="Department", hue="Attrition", data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"],
             bins=15,
             kde=True,
             color="#6A5ACD")

plt.title("Distribution of Employee Age", fontsize=14, fontweight="bold")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(y="Department", data=df, color="red")

plt.title("Number of Employees by Department")
plt.xlabel("Number of Employees")
plt.ylabel("Department")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Attrition",
    y="MonthlyIncome",
    data=df,
    palette=["#4CAF50", "#FF4D4D"]
)

plt.title("Monthly Income vs Attrition")
plt.xlabel("Attrition")
plt.ylabel("Monthly Income")

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="EducationField",
    data=df,
    color="red"
)

plt.title("Employees by Education Field")
plt.xlabel("Education Field")
plt.ylabel("Number of Employees")

plt.xticks(rotation=20)

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="BusinessTravel",
    data=df,
    color="orange"
)

plt.title("Business Travel Frequency")
plt.xlabel("Business Travel")
plt.ylabel("Number of Employees")

plt.show()
```
